```python
def real_time_detection():
    # on va utiliser la webcam de l'ordinateur pour détecter les visages en temps réel
    import cv2
    cap = cv2.VideoCapture(0)
    
    car_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
    
    while True:
        # reads frames from a video
        ret, frames = cap.read()
        
        # convert to gray scale of each frames
        gray = cv2.cvtColor(frames, cv2.COLOR_BGR2GRAY)
        
        # Detects cars of different sizes in the input image
        cars = car_cascade.detectMultiScale( gray, 1.1, 1)

        # Determination du fragment d'image ayant la plus grande superfice

        # calcul des superficies
        if len(cars) > 1:
            areas = []
            indices = []
            for (x,y,w,h), number in zip(cars, range(0, len(cars))):
                areas.append(w*h)
                indices.append(number)

            # identification de l'indice du fragement ayant la plus grande superficie
            import numpy as np
            areas = np.array(areas)
            indices = np.array(indices)
            maxi = np.max(areas)
            ind = np.where(areas == maxi)
            goodIndice = indices[ind]

            # selection du fragment
            goodImage = cars[goodIndice]
        
            # To draw a rectangle in each cars
            for (x,y,w,h) in cars[goodIndice]:
                cv2.rectangle(frames,(x,y),(x+w,y+h),(0,0,255),2)
                font = cv2.FONT_HERSHEY_DUPLEX
                cv2.putText(frames, 'Visage', (x + 6, y - 6), font, 0.5, (0, 0, 255), 1)
                # Display frames in a window
                cv2.imshow('Detection Visage', frames)

        elif len(cars) == 1:
            # To draw a rectangle in each cars
            for (x,y,w,h) in cars:
                cv2.rectangle(frames,(x,y),(x+w,y+h),(0,0,255),2)
                font = cv2.FONT_HERSHEY_DUPLEX
                cv2.putText(frames, 'Visage', (x + 6, y - 6), font, 0.5, (0, 0, 255), 1)
                # Display frames in a window
                cv2.imshow('Detection Visage', frames)

        elif len(cars) == 0:
            # Display frames in a window
            cv2.imshow('Detection Visage', frames)
            
        # Wait for Enter key to stop
        if cv2.waitKey(33) == 13:
            break

    cap.release()
    cv2.destroyAllWindows()

real_time_detection()
```
