                    Modèle pré entraîné et détection d'objet dans une image

```python
# importation librairies
import cv2
from skimage.io import imread, imshow
```

# Exemple 1

```python
# charger l'image
img = cv2.imread('image.jpg')
imshow(img)
```

```python
# passage du BGR au RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
imshow(img_rgb)
```

```python
# charger le modèle de detection de visage
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
# appliquer le modele de detection à l'image pour detecter le(s) visages
faces = face_cascade.detectMultiScale(img, 1.1, 4)
faces
```

```python
len(faces)
```

```python
# Representer un contour sur le ou les visages détectés

for (x, y, w, h), number in zip( faces, range(0, len(faces))  ) : 
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)
    name = "Face detected"+str(number)+".png"
    cv2.imwrite(name, img)
    print('Enregistré avec succès')
```

```python
# affichage visages détectés

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
imshow(img_rgb)
```

## Definir une fonction de détection de visage

```python
def faceDetection(picture):
    # charger l'image
    img = cv2.imread(picture)
    
    # charger le modèle de detection de visage
    face= cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
    
    # appliquer le modele de detection à l'image pour detecter le(s) visages
    faces= face.detectMultiScale(img, 1.1, 4)
    
    # Representer un contour sur le ou les visages détectés
    
    for (x, y, w, h), number in zip( faces, range(0,len(faces)) ):
        cv2.rectangle(img, (x,y), (x+w, y+h), (0,230+number,0), 2)
        name= f"Face {number} dectected.png"
        cv2.imwrite(name, img)
        print('Image enregistrée avec succès')
        
    return img
```

```python
# application à l'image précédente

# charger l'image
img = 'image.jpg'

# application de la fonction de detection
output = faceDetection(img)

# affichage visages détectés
output_rgb = cv2.cvtColor(output, cv2.COLOR_BGR2RGB)
imshow(output_rgb)
```

## Exemple 2

```python
# Charger l'image
img2= "image2.jpg"

# application de la fonction de detection
output = faceDetection(img2)

# affichage visages détectés
output_rgb = cv2.cvtColor(output, cv2.COLOR_BGR2RGB)
imshow(output_rgb)
```
