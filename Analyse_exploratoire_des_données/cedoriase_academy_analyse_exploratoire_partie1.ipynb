### 6 Regroupement

En général, le regroupement des données dans Pandas fonctionne comme suit :

df.groupby(by=colonnes_de_regroupement)[colonnes_a_afficher].fonction()

- Tout d'abord, la méthode groupby divise les colonnes_de_regroupement par leurs valeurs. Ils deviennent un nouvel index dans le dataframe résultant.


- Ensuite, les colonnes d'intérêt sont sélectionnées (colonnes_a_afficher). 


- Si colonnes_a_afficher n'est pas inclus, toutes les clauses non groupby seront incluses.
     
     
- Enfin, une ou plusieurs fonctions sont appliquées aux groupes obtenus par colonnes sélectionnées.

##### Voici un exemple où nous regroupons les données en fonction des valeurs de la variable Churn et affichons les statistiques de trois colonnes dans chaque groupe :

```python
# Exemple 1 : Fonction 'moyenne' pour la colonne de regroupement

colonnes= ["Total day minutes", "Total eve minutes"]
df.groupby( ["Churn"] )[colonnes].mean()
```

```python
# Exemple 2 : statistiques descriptives pour la colonne de regroupement
df.groupby("Churn").agg({"Total intl minutes": "mean"})
# agg permet de calculer plusieurs statistiques descriptives sur une ou plusieurs 
# colonnes du DataFrame, en fonction d'une ou plusieurs colonnes de regroupement.
```

- Interprétation : les abonnés résiliés ont en moyenne consommé plus de minutes internationales que les abonnés non résiliés.

```python
# Exemple 3 : Faisons la même chose, mais légèrement différemment en passant une liste de fonctions à agg() :

df.groupby("Churn").agg({"Total intl minutes": ["mean", "std"]})
```

- Interprétation : les abonnés résiliés ont en moyenne consommé plus de minutes internationales que les abonnés non résiliés, et les abonnés résiliés ont une plus grande variabilité dans leur consommation de minutes internationales que les abonnés non résiliés.

### 8 Tableaux croisés dynamiques 

Ils sont implémentés dans Pandas avec la méthode pivot_table avec les paramètres suivants :

- values : une liste de variables pour lesquelles calculer des statistiques,
     
- index : une liste de variables pour regrouper les données,

- aggfunc : quelles statistiques nous devons calculer pour les groupes, ex. somme, moyenne, maximum, minimum ou autre chose.

##### Exemple 1 : affichons le nombre total d'abonnés qui effectuent des appels le jour, le soir et la nuit par indicatif régional :

```python
df.pivot_table( ["Total day calls", "Total eve calls", "Total night calls"], 
                ["Area code"], 
                aggfunc= "sum" )
```

##### Exemple 2 : affichons le nombre moyen journalier d'abonnés qui effectuent des appels le jour, le soir et la nuit par indicatif régional :


```python
df.pivot_table(
     values=["Total day calls", "Total eve calls", "Total night calls"],
     index="Area code",
     aggfunc="mean"
)
```

### 9 Transformations de DataFrame

L'ajout de colonnes à un DataFrame est possible.

Par exemple, si nous voulons calculer le nombre total d'appels effectués par tous les abonnés, créons la série total_calls et collons-la au DataFrame :

```python
# # Etape 1 : Création de la série total_calls
total_calls = (
     df["Total day calls"]
     + df["Total eve calls"]
     + df["Total night calls"]
     + df["Total intl calls"]
)

total_calls.head()
```

```python
# Etape 2 : Ajout de la colonne "Total calls" à la fin du DataFrame

# le paramètre loc est le nombre de colonnes après lesquelles nous insérons les nouvelles variables
# nous lui attribuons la valeur len(df.columns) pour le coller à la fin du dataframe

df.insert(loc=len(df.columns), column="Total calls", value=total_calls)

df.head()
```

- Il est possible d'ajouter directement une colonne au tableau sans créer la variable au préalable :

```python
# Ajouter directement au DataFrame la somme des charges
df["Total charge"] = (
     df["Total day charge"]
     + df["Total eve charge"]
     + df["Total night charge"]
     + df["Total intl charge"]
)

df.head()
```

### 10 Suppression de lignes ou de colonnes

Pour supprimer des colonnes ou des lignes :

- utilisez la méthode drop, en passant les index requis et le paramètre d'axe (1 si vous supprimez des colonnes, et rien ou 0 si vous supprimez des lignes). 


- L'argument inplace indique s'il faut modifier le DataFrame d'origine. 


- Avec inplace=False, la méthode drop ne modifie pas le DataFrame existant et en renvoie un nouveau avec des lignes ou des colonnes supprimées. 


- Avec inplace=True, cela modifie le DataFrame.

```python
# suppression des colonnes récemment créées ("Total charge", "Total calls")
df.drop(columns=["Total charge", "Total calls"], inplace=True)
df.head(5)
```

```python
# suppression des lignes ayant les indices de position 1 et 2
df.drop(index=[1, 2], inplace=True)
df.head(5)
```
